Reordering the city and day axes by their totals moves the strongest cells together, making the cluster easier to detect. ## Task 3 — Will the Slider Still Feel Instant? (Request #3)

**Real-world usage:** Interactive tools are sized against latency budgets before they're built.

**Why this technique:** **Dynamic queries** need feedback within ~0.1 s (Slides 15–16).

**TODO:**
1. Build a 1-million-row table by repeating `tx`, and measure the time of the filter
   `(detector_score >= 0.6) & (amount_inr > 3000)` (take the best of 5 runs).
2. Compute the per-item cost (µs) and the maximum number of rows filterable within 0.1 s on this machine.
3. Estimate the response time for 50,000,000 rows and the % of rows that would need to be skipped (pre-indexed /
   pre-aggregated) to meet the 0.1 s budget.

```python
big = pd.concat([tx] * 182, ignore_index=True)

# TODO 1: best of five filter timings on approximately one million rows
timings = []
for _ in range(5):
    start = time.perf_counter()
    _ = big[(big["detector_score"] >= 0.6) & (big["amount_inr"] > 3000)]
    timings.append(time.perf_counter() - start)
best_s = min(timings)

# TODO 2: per-item cost and 0.1-second capacity
per_item_us = (best_s / len(big)) * 1_000_000
max_rows = 0.1 / (best_s / len(big))

# TODO 3: linear estimate for 50 million rows
t_50m = (best_s / len(big)) * 50_000_000
skip_pct = max(0, (1 - 0.1 / t_50m) * 100)

print(f"{len(big):,} rows in {best_s*1000:.1f} ms -> {per_item_us:.4f} us/row -> max {max_rows:,.0f} rows in 0.1 s")
print(f"50M rows -> {t_50m:.2f} s; must skip {skip_pct:.1f}% of rows to stay under 0.1 s")
print("Timing is hardware- and load-dependent; this is a linear estimate, not a production benchmark.")
```

**Reflection — Answer:**

A **sampling** or **pre-aggregation** strategy can reduce the number of records scanned on every slider movement. For example, the interface could first show aggregated counts by time bucket and category, then query the underlying transactions only after the analyst narrows the time range. The trade-off is loss of transaction-level detail in the initial view; rare cases and small clusters may be hidden or diluted by aggregation.

## Task 4 — Steer the Threshold to the Team's Capacity (Request #4)

**Real-world usage:** Alert thresholds are set against investigator capacity, not just statistical optimality.

**Why this technique:** **Human-in-the-loop** steering of a model parameter (Slide 37) evaluated with **precision, recall,
F1** (Slides 40–41).

**TODO:**
1. For thresholds 0.05, 0.10, …, 0.95 compute alerts (flagged count), TP, FP, FN, precision, recall and F1.
2. Find the threshold with the **highest recall** among those producing **≤ 150 alerts**. Also find the best-F1 threshold.
3. Plot the precision–recall curve and mark the current threshold (0.5), the capacity-constrained choice and the best-F1
   choice.
4. Print a one-line recommendation.

```python
rows = []
for th in np.round(np.arange(0.05, 1.00, 0.05), 2):
    flagged_at_th = tx["detector_score"] >= th
    tp = int((flagged_at_th & (tx["is_fraud"] == 1)).sum())
    fp = int((flagged_at_th & (tx["is_fraud"] == 0)).sum())
    fn = int((~flagged_at_th & (tx["is_fraud"] == 1)).sum())
    precision = tp / (tp + fp) if tp + fp else 0
    recall = tp / (tp + fn) if tp + fn else 0
    f1 = 2 * precision * recall / (precision + recall) if precision + recall else 0
    rows.append(dict(threshold=th, alerts=tp + fp, TP=tp, FP=fp, FN=fn,
                     precision=precision, recall=recall, F1=f1))

pr = pd.DataFrame(rows)

# TODO 2: highest recall under the 150-alert capacity and best F1 overall
feasible = pr[pr["alerts"] <= 150]
capacity_choice = feasible.sort_values(["recall", "threshold"], ascending=[False, True]).iloc[0]
best_f1 = pr.loc[pr["F1"].idxmax()]

# TODO 3: precision-recall curve with the three operating points
fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(pr["recall"], pr["precision"], "o-", color="#4C72B0", label="Threshold sweep")
for row, label, color in [
    (pr.loc[(pr.threshold - 0.5).abs().idxmin()], "Current 0.50", "#777777"),
    (capacity_choice, f"Capacity choice {capacity_choice.threshold:.2f}", "#D9534F"),
    (best_f1, f"Best F1 {best_f1.threshold:.2f}", "#2E8B57")
]:
    ax.scatter(row["recall"], row["precision"], s=100, color=color, label=label, zorder=3)
ax.set_xlabel("Recall"); ax.set_ylabel("Precision")
ax.set_title("Precision–recall trade-off across detector thresholds")
ax.grid(alpha=0.25); ax.legend()

plt.tight_layout(); plt.show()

# TODO 4: operational recommendation
print(f"Recommend threshold {capacity_choice.threshold:.2f}: {int(capacity_choice.alerts)} alerts, "
      f"precision {capacity_choice.precision:.1%}, recall {capacity_choice.recall:.1%}. "
      f"Best-F1 threshold is {best_f1.threshold:.2f} (F1 {best_f1.F1:.3f}); "
      "the Head of Fraud Ops should approve the operating point.")
```

**Reflection — Answer:**

The **Head of Fraud Ops** should make the final decision with the model providing evidence, because the threshold is an operational trade-off involving investigator capacity and the cost of missed fraud. The capacity-constrained choice is **0.80**, which produces **144 alerts**, with **58.3% precision** and **70.0% recall**. The best-F1 threshold is **0.85**, with F1 approximately **0.637**, but it produces a different balance of false positives and missed cases. Visual analytics keeps a human in the loop so the team can inspect the trade-off, apply domain knowledge and remain accountable for the decision (Slides 35–37).

## Task 6 — Prove or Kill the "Fraud Ring" Hypothesis (Request #6)

**Real-world usage:** Before escalating to law enforcement, the team must rule out simpler explanations.

**Why this technique:** **Hypothesis testing** with visual + statistical confirmation and an explicit **counter-example
search** to guard against confirmation bias (Slides 42–43).

**TODO:**
1. State the hypothesis precisely using what you found in Task 1 (city and day range).
2. Compute the fraud rate inside vs. outside that city–day window and a chi-square p-value
   (`scipy.stats.chi2_contingency` on `pd.crosstab(in_window, tx.is_fraud)`).
3. Counter-example searches: fraud rate for (a) the same city on other days, (b) other cities on the same days.
4. Visual confirmation: a bar chart of the four groups (window / same city other days / other cities same days / rest).
5. Print your verdict.

```python
from scipy.stats import chi2_contingency

# TODO 1: hypothesis from the reordered overview — Pune, days 18–20
in_window = (tx["city"] == "Pune") & tx["day"].between(18, 20)

# TODO 2: compare fraud rates and run chi-square test
contingency = pd.crosstab(in_window, tx["is_fraud"])
chi2, p_value, dof, expected = chi2_contingency(contingency)
window_rate = tx.loc[in_window, "is_fraud"].mean()
outside_rate = tx.loc[~in_window, "is_fraud"].mean()
print(f"Hypothesis: an unusual fraud concentration occurred in Pune on days 18–20.")
print(f"Window: {in_window.sum()} transactions; fraud rate {window_rate:.1%}")
print(f"Outside window: {(~in_window).sum()} transactions; fraud rate {outside_rate:.1%}")
print(f"Chi-square = {chi2:.2f}, df = {dof}, p-value = {p_value:.3e}")

# TODO 3: counter-example searches
same_city_other_days = tx[(tx["city"] == "Pune") & ~tx["day"].between(18, 20)]
other_cities_same_days = tx[(tx["city"] != "Pune") & tx["day"].between(18, 20)]
rest = tx[(tx["city"] != "Pune") & ~tx["day"].between(18, 20)]
print(f"Pune on other days: {same_city_other_days.is_fraud.mean():.1%} "
      f"({same_city_other_days.is_fraud.sum()} / {len(same_city_other_days)})")
print(f"Other cities on days 18–20: {other_cities_same_days.is_fraud.mean():.1%} "
      f"({other_cities_same_days.is_fraud.sum()} / {len(other_cities_same_days)})")

# TODO 4: visual comparison of the four groups
group_data = {
    "Pune, days 18–20": tx.loc[in_window, "is_fraud"],
    "Pune, other days": same_city_other_days["is_fraud"],
    "Other cities, days 18–20": other_cities_same_days["is_fraud"],
    "Rest": rest["is_fraud"]
}
group_rates = {name: values.mean() * 100 for name, values in group_data.items()}
fig, ax = plt.subplots(figsize=(9, 4))
bars = ax.bar(group_rates.keys(), group_rates.values(),
              color=["#D9534F", "#999999", "#999999", "#999999"])
ax.set_ylabel("Confirmed fraud rate (%)")
ax.set_title("Counter-example search: fraud rate across four groups")
ax.tick_params(axis="x", rotation=18)
for bar, name in zip(bars, group_data):
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height(),
            f"{group_rates[name]:.1f}%\n(n={len(group_data[name])})",
            ha="center", va="bottom", fontsize=8)
plt.tight_layout(); plt.show()

# TODO 5: verdict based on effect size, statistical evidence and counter-examples
if p_value < 0.05 and window_rate > same_city_other_days["is_fraud"].mean() and window_rate > other_cities_same_days["is_fraud"].mean():
    verdict = ("The data support a localized Pune, days 18–20 fraud hotspot: its rate is substantially higher "
               "than both counter-example groups and the association is statistically significant. "
               "This is evidence of a cluster, not proof that a coordinated fraud ring caused it.")
else:
    verdict = ("The evidence does not establish a distinct Pune, days 18–20 hotspot; investigate alternative explanations.")
print("VERDICT:", verdict)
```

**Reflection — Answer:**

The **Pune on other days** comparison would challenge the idea of a time-limited ring if its fraud rate were similarly high; that would support the alternative explanation that Pune has a generally higher fraud rate. The **other cities on days 18–20** comparison would challenge the city-specific part if those cities also showed a similar spike; that would suggest a broader event or system-wide issue on those dates. Both counter-examples matter because they test whether the pattern is specific to the proposed city–day combination.
